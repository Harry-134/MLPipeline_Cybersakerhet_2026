```python
# Read training data from Catalog
df_train = spark.table("workspace.default.kdd_train_filtered")

# Display training data information
print("TRAINING DATA")
print("-" * 40)
print(f"Rows: {df_train.count()}")
print(f"Columns: {len(df_train.columns)}")

print("\nSchema:")
df_train.printSchema()

print("\nSample (first 5 rows):")
display(df_train.limit(5))
```

```python
# Read test data from Catalog
df_test = spark.table("workspace.default.kdd_test_filtered")

# Display test data information
print("TEST DATA")
print("-" * 40)
print(f"Rows: {df_test.count()}")
print(f"Columns: {len(df_test.columns)}")

print("\nSchema:")
df_test.printSchema()

print("\nSample (first 5 rows):")
display(df_test.limit(5))
```

```python
# Convert Spark DataFrames to Pandas for scikit-learn compatibility
pandas_train = df_train.toPandas()
pandas_test = df_test.toPandas()

# Display dataset dimensions
print("DATA CONVERSION COMPLETE")
print("-" * 40)
print(f"Training data: {pandas_train.shape[0]} rows, {pandas_train.shape[1]} columns")
print(f"Test data: {pandas_test.shape[0]} rows, {pandas_test.shape[1]} columns")
```

```python
# Separate features (X) from target variable (y)
# The target column is always the last column in NSL-KDD dataset
X_train = pandas_train.iloc[:, :-1]
y_train = pandas_train.iloc[:, -1]

X_test = pandas_test.iloc[:, :-1]
y_test = pandas_test.iloc[:, -1]

print("FEATURES AND TARGET SEPARATED")
print("-" * 40)
print(f"Features (X_train): {X_train.shape[1]} columns")
print(f"Target variable (y_train): {y_train.name}")
print(f"Features (X_test): {X_test.shape[1]} columns")
print(f"Target variable (y_test): {y_test.name}")
```

```python
# Import required libraries for machine learning
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
import mlflow
import mlflow.sklearn
import joblib

print("LIBRARIES IMPORTED")
print("-" * 40)
print("pandas, numpy, scikit-learn, mlflow, joblib loaded successfully")
```

```python
# Identify categorical columns (text columns)
categorical_columns = X_train.select_dtypes(include=['object']).columns.tolist()

print("CATEGORICAL COLUMNS FOUND")
print("-" * 40)
print(f"Columns to encode: {categorical_columns}")

# Apply one-hot encoding to categorical columns
X_train_encoded = pd.get_dummies(X_train, columns=categorical_columns, drop_first=True)
X_test_encoded = pd.get_dummies(X_test, columns=categorical_columns, drop_first=True)

# Ensure test data has the same columns as training data
X_test_encoded = X_test_encoded.reindex(columns=X_train_encoded.columns, fill_value=0)

print("\nENCODING COMPLETE")
print("-" * 40)
print(f"Training features: {X_train_encoded.shape[1]} columns")
print(f"Test features: {X_test_encoded.shape[1]} columns")
```

```python
# Convert target variable from text to binary (0 = normal, 1 = attack)
def encode_target(value):
    """Convert target value to binary classification."""
    if str(value).lower() != 'normal':
        return 1
    return 0

y_train = y_train.apply(encode_target)
y_test = y_test.apply(encode_target)

# Verify class distribution
print("TARGET ENCODING COMPLETE")
print("-" * 40)
print("\nTraining set class distribution:")
print(y_train.value_counts())
print("\nTest set class distribution:")
print(y_test.value_counts())

# Combine training and test data for better evaluation
X_combined = pd.concat([X_train_encoded, X_test_encoded], axis=0)
y_combined = pd.concat([y_train, y_test], axis=0)

print("\nCOMBINED DATA")
print("-" * 40)
print(f"Combined features: {X_combined.shape[0]} rows, {X_combined.shape[1]} columns")
print(f"Combined target: {y_combined.value_counts().to_dict()}")

# Split combined data into new train/test sets (80/20)
X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(
    X_combined, y_combined, test_size=0.2, random_state=42
)

print(f"\nNew training data: {X_train_final.shape[0]} rows")
print(f"New test data: {X_test_final.shape[0]} rows")
```

```python
# Initialize and train the model with combined data
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_final, y_train_final)

print("MODEL TRAINING COMPLETE")
print("-" * 40)
print("Random Forest model trained successfully")
```

```python
# Make predictions on test data
y_pred = rf_model.predict(X_test_final)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test_final, y_pred)
precision = precision_score(y_test_final, y_pred)
recall = recall_score(y_test_final, y_pred)

# Display results
print("MODEL PERFORMANCE METRICS")
print("-" * 40)
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")

print("\nConfusion Matrix:")
cm = confusion_matrix(y_test_final, y_pred)
print(f"[[{cm[0][0]:4d} {cm[0][1]:4d}]")
print(f" [{cm[1][0]:4d} {cm[1][1]:4d}]]")

print("\nDetailed Classification Report:")
print(classification_report(y_test_final, y_pred, target_names=['Normal', 'Attack']))
```

```python
# Log model, parameters, and metrics to MLflow
with mlflow.start_run() as run:
    # Log parameters
    mlflow.log_param("model_type", "RandomForestClassifier")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("random_state", 42)
    
    # Log metrics
    mlflow.log_metric("accuracy", accuracy)
    mlflow.log_metric("precision", precision)
    mlflow.log_metric("recall", recall)
    
    # Log the model
    mlflow.sklearn.log_model(rf_model, "random_forest_model")
    
    print("MLFLOW EXPERIMENT LOGGED")
    print("-" * 40)
    print(f"Run ID: {run.info.run_id}")
    print("Parameters, metrics, and model saved to MLflow")
```

```python
# Save model as pickle file for GitHub upload
model_filename = "random_forest_model.pkl"
joblib.dump(rf_model, model_filename)

print("MODEL SAVED")
print("-" * 40)
print(f"Model saved as: {model_filename}")
print("Download the file from Workspace -> right-click -> Download")
```

```python
# Test model with a sample from the test set
sample_index = 0
sample_features = X_test_final.iloc[sample_index:sample_index+1]
sample_actual = y_test_final.iloc[sample_index]
sample_prediction = rf_model.predict(sample_features)[0]

print("SAMPLE PREDICTION TEST")
print("-" * 40)
print(f"Actual:    {'Attack' if sample_actual == 1 else 'Normal'}")
print(f"Predicted: {'Attack' if sample_prediction == 1 else 'Normal'}")

if sample_actual == sample_prediction:
    print("Prediction correct!")
else:
    print("Prediction incorrect")
```
